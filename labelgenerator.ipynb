```python
import json, os

TRAIN_DIR = "/kaggle/input/fruits/fruits-360_100x100/fruits-360/Training"

classes = sorted([d for d in os.listdir(TRAIN_DIR) if os.path.isdir(os.path.join(TRAIN_DIR, d))])
labels_map = {i: classes[i] for i in range(len(classes))}

with open("/kaggle/working/labels_map.json", "w") as f:
    json.dump(labels_map, f)

print("Created labels_map.json with", len(labels_map), "classes.")
```
